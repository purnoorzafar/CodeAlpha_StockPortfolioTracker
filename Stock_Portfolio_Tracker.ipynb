```python
from IPython.display import display, HTML, clear_output

# ============================================================
# CODEALPHA TASK 2
# STOCK PORTFOLIO TRACKER
# ============================================================

# ---------------- STOCK PRICE DATA ----------------

stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "MSFT": 420,
    "AMZN": 175,
    "META": 500,
    "NFLX": 650,
    "NVDA": 120
}

# Portfolio list
portfolio = []


# ============================================================
# DISPLAY AVAILABLE STOCKS
# ============================================================

print("=" * 60)
print("              STOCK PORTFOLIO TRACKER")
print("=" * 60)

print("\nAvailable Stocks:\n")

for stock, price in stock_prices.items():
    print(f"  {stock:<8} ${price:,.2f}")


# ============================================================
# ENTER PORTFOLIO DATA
# ============================================================

print("\n" + "=" * 60)
print("              ENTER YOUR STOCK DATA")
print("=" * 60)

print("\nType 'done' when you have finished entering stocks.\n")


while True:

    stock = input("Enter stock symbol: ").upper().strip()

    # Finish
    if stock == "DONE":
        break

    # Check stock
    if stock not in stock_prices:

        print("\n❌ Stock not available.")

        print(
            "Available stocks:",
            ", ".join(stock_prices.keys())
        )

        print()

        continue

    # Quantity
    try:

        quantity = int(
            input(f"Enter quantity for {stock}: ")
        )

        if quantity <= 0:

            print(
                "❌ Quantity must be greater than zero.\n"
            )

            continue

    except ValueError:

        print(
            "❌ Please enter a valid number.\n"
        )

        continue

    # Price
    price = stock_prices[stock]

    # Calculate value
    total_value = price * quantity

    # Add to portfolio
    portfolio.append({
        "Stock": stock,
        "Quantity": quantity,
        "Price": price,
        "Total Value": total_value
    })

    print("\n✅ Stock added successfully!")

    print(
        f"   {stock} × {quantity} = "
        f"${total_value:,.2f}"
    )

    print()


# ============================================================
# DASHBOARD
# ============================================================

clear_output(wait=True)


if portfolio:

    # --------------------------------------------------------
    # CALCULATIONS
    # --------------------------------------------------------

    total_investment = sum(
        item["Total Value"]
        for item in portfolio
    )

    total_shares = sum(
        item["Quantity"]
        for item in portfolio
    )

    total_holdings = len(portfolio)

    average_value = (
        total_investment / total_holdings
    )


    # --------------------------------------------------------
    # CREATE TABLE ROWS
    # --------------------------------------------------------

    rows = ""

    for item in portfolio:

        rows += f"""
        <tr>

            <td>
                <span class="stock-badge">
                    {item['Stock']}
                </span>
            </td>

            <td>
                {item['Quantity']}
            </td>

            <td>
                ${item['Price']:,.2f}
            </td>

            <td>
                <strong>
                    ${item['Total Value']:,.2f}
                </strong>
            </td>

        </tr>
        """


    # ========================================================
    # PROFESSIONAL DASHBOARD
    # ========================================================

    dashboard = f"""

    <style>

    /* MAIN DASHBOARD */

    .dashboard {{

        font-family:
            Arial,
            Helvetica,
            sans-serif;

        max-width: 1100px;

        margin: 30px auto;

        background: #f4f7f6;

        padding: 30px;

        border-radius: 24px;

        box-shadow:
            0 10px 35px
            rgba(0,0,0,0.10);

    }}


    /* HEADER */

    .header {{

        background:
            linear-gradient(
                135deg,
                #063b25,
                #198754
            );

        color: white;

        padding: 35px;

        border-radius: 20px;

        text-align: center;

        margin-bottom: 25px;

    }}


    .header h1 {{

        margin: 0;

        font-size: 34px;

        letter-spacing: 0.5px;

    }}


    .header p {{

        margin-top: 10px;

        font-size: 15px;

        opacity: 0.9;

    }}


    /* STATUS */

    .status {{

        display: inline-block;

        margin-top: 15px;

        padding: 8px 18px;

        border-radius: 20px;

        background: rgba(255,255,255,0.15);

        font-size: 13px;

    }}


    /* CARDS */

    .cards {{

        display: grid;

        grid-template-columns:
            repeat(4, 1fr);

        gap: 18px;

        margin-bottom: 25px;

    }}


    .card {{

        background: white;

        padding: 24px;

        border-radius: 18px;

        box-shadow:
            0 5px 18px
            rgba(0,0,0,0.07);

        text-align: center;

        transition: 0.3s;

    }}


    .card:hover {{

        transform: translateY(-5px);

        box-shadow:
            0 10px 25px
            rgba(0,0,0,0.12);

    }}


    .card-icon {{

        font-size: 27px;

        margin-bottom: 8px;

    }}


    .card-title {{

        color: #6c757d;

        font-size: 12px;

        font-weight: bold;

        letter-spacing: 1px;

    }}


    .card-value {{

        margin-top: 10px;

        color: #198754;

        font-size: 25px;

        font-weight: bold;

    }}


    /* PORTFOLIO SECTION */

    .section {{

        background: white;

        padding: 28px;

        border-radius: 18px;

        box-shadow:
            0 5px 18px
            rgba(0,0,0,0.06);

    }}


    .section-header {{

        display: flex;

        justify-content:
            space-between;

        align-items: center;

        margin-bottom: 15px;

    }}


    .section h2 {{

        margin: 0;

        color: #063b25;

        font-size: 22px;

    }}


    .section-label {{

        background: #e8f5ee;

        color: #198754;

        padding: 7px 13px;

        border-radius: 20px;

        font-size: 12px;

        font-weight: bold;

    }}


    /* TABLE */

    table {{

        width: 100%;

        border-collapse:
            separate;

        border-spacing: 0;

        overflow: hidden;

        margin-top: 15px;

    }}


    th {{

        background: #198754;

        color: white;

        padding: 15px;

        text-align: left;

        font-size: 14px;

    }}


    td {{

        padding: 16px;

        border-bottom:
            1px solid #eeeeee;

        font-size: 14px;

    }}


    tr:hover td {{

        background: #f4faf7;

    }}


    /* STOCK BADGE */

    .stock-badge {{

        display: inline-block;

        background: #e8f5ee;

        color: #0f5132;

        padding: 7px 13px;

        border-radius: 8px;

        font-weight: bold;

    }}


    /* FOOTER */

    .footer {{

        text-align: center;

        margin-top: 25px;

        color: #6c757d;

        font-size: 13px;

    }}


    .footer strong {{

        color: #198754;

    }}


    /* RESPONSIVE */

    @media (max-width: 800px) {{

        .cards {{

            grid-template-columns:
                repeat(2, 1fr);

        }}

    }}


    @media (max-width: 500px) {{

        .cards {{

            grid-template-columns: 1fr;

        }}

        .dashboard {{

            padding: 15px;

        }}

        .section {{

            overflow-x: auto;

        }}

    }}

    </style>


    <!-- ================================================== -->
    <!-- DASHBOARD -->
    <!-- ================================================== -->

    <div class="dashboard">


        <!-- HEADER -->

        <div class="header">

            <h1>
                📈 Stock Portfolio Tracker
            </h1>

            
            <div class="status">
                ● Portfolio Active
            </div>

        </div>


        <!-- ================================================== -->
        <!-- SUMMARY CARDS -->
        <!-- ================================================== -->

        <div class="cards">


            <!-- TOTAL INVESTMENT -->

            <div class="card">

                <div class="card-icon">
                    💰
                </div>

                <div class="card-title">
                    TOTAL INVESTMENT
                </div>

                <div class="card-value">
                    ${total_investment:,.2f}
                </div>

            </div>


            <!-- STOCK HOLDINGS -->

            <div class="card">

                <div class="card-icon">
                    📊
                </div>

                <div class="card-title">
                    STOCK HOLDINGS
                </div>

                <div class="card-value">
                    {total_holdings}
                </div>

            </div>


            <!-- TOTAL SHARES -->

            <div class="card">

                <div class="card-icon">
                    📦
                </div>

                <div class="card-title">
                    TOTAL SHARES
                </div>

                <div class="card-value">
                    {total_shares}
                </div>

            </div>


            <!-- AVERAGE VALUE -->

            <div class="card">

                <div class="card-icon">
                    📈
                </div>

                <div class="card-title">
                    AVG. HOLDING VALUE
                </div>

                <div class="card-value">
                    ${average_value:,.2f}
                </div>

            </div>


        </div>


        <!-- ================================================== -->
        <!-- PORTFOLIO TABLE -->
        <!-- ================================================== -->

        <div class="section">


            <div class="section-header">

                <h2>
                    📋 My Portfolio
                </h2>

                <div class="section-label">
                    {total_holdings} Holdings
                </div>

            </div>


            <table>

                <thead>

                    <tr>

                        <th>
                            Stock
                        </th>

                        <th>
                            Quantity
                        </th>

                        <th>
                            Price
                        </th>

                        <th>
                            Total Value
                        </th>

                    </tr>

                </thead>


                <tbody>

                    {rows}

                </tbody>

            </table>


        </div>


        <!-- ================================================== -->
        <!-- FOOTER -->
        <!-- ================================================== -->

        <div class="footer">

            Stock Portfolio Tracker

            <br>

            Developed for
            <strong>
                CodeAlpha Python Programming Internship
            </strong>

        </div>


    </div>

    """


    # DISPLAY DASHBOARD

    display(
        HTML(dashboard)
    )


else:

    # ========================================================
    # EMPTY PORTFOLIO
    # ========================================================

    display(
        HTML("""

        <div style="
            max-width:800px;
            margin:40px auto;
            padding:30px;
            background:#fff3cd;
            border-radius:18px;
            text-align:center;
            font-family:Arial;
            color:#856404;
        ">

            <h2>
                ⚠️ No Portfolio Data
            </h2>

            <p>
                You have not added any stocks yet.
            </p>

        </div>

        """)
    )
```

```python
# ============================================================
# SAVE PORTFOLIO TO CSV
# ============================================================

import csv

if portfolio:

    filename = "my_stock_portfolio.csv"

    with open(filename, "w", newline="", encoding="utf-8") as file:

        writer = csv.DictWriter(
            file,
            fieldnames=[
                "Stock",
                "Quantity",
                "Price",
                "Total Value"
            ]
        )

        writer.writeheader()
        writer.writerows(portfolio)

    print("✅ Portfolio saved successfully!")
    print(f"📁 File name: {filename}")

else:

    print("⚠️ No portfolio data available.")
```
